# Data Cleaning

```python
#finding the duplicated values
df.duplicated().sum()
```

```python
#dropping the duplicated values
df.drop_duplicates(inplace=True)
```

```python
#finding the null data
df.isnull().sum()
```

```python
#dropping the null data
df.dropna(inplace=True)
```

```python
#data transformation
df['unit_price'] = df['unit_price'].str.replace("$", "").astype(float)
df.head()
```

```python
#creating new column 
df['total'] = df['unit_price'] * df['quantity']
df.head()
```

```python
df.columns = df.columns.str.lower()
```

```python
from sqlalchemy import create_engine

# Update with your SQL Server details
server = 'DESKTOP-3Q2DPDN\SQLEXPRESS'
database = 'walmart_sales_data'

# Create the SQLAlchemy engine for MSSQL
# (Using Windows/Trusted Authentication in this example)
engine = create_engine(f"mssql+pyodbc://@{server}/{database}?driver=ODBC+Driver+17+for+SQL+Server&trusted_connection=yes")
table_name = 'walmart'

# Test the connection by pulling data directly into a Pandas DataFrame
df.to_sql(table_name, engine, if_exists='replace', index=False)

print(f"Data successfully loaded into table '{table_name}' in database '{database}'.")
```
